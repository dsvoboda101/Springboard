```python
meta = pd.read_csv('68debdfdd1b8e9f8fd64dab0_sea-ad_cohort_donor_metadata_072524.csv')
```

```python
meta_short = meta[['Donor ID','APOE Genotype', 'Cognitive Status', 'Known head injury','Consensus Clinical Dx (choice=Alzheimers disease)', 
                   'Consensus Clinical Dx (choice=Taupathy)', 'Last CASI Score', 'Last MMSE Score', 'Overall AD neuropathological Change', 'Thal', 'Braak', 'Overall CAA Score', 
                   'CERAD score','Severely Affected Donor']].copy()
meta_short.isna().sum()
```

```python
meta_short['Severely Affected Donor'] = meta_short['Severely Affected Donor'].fillna('No')
meta_short['Known head injury'] = meta_short['Known head injury'].fillna('Unknown')
meta_short['Severely Affected Donor'].value_counts()
```

The quantification of each marker is based on their unique patterns, so feature engineering and analysis cannot be standarized accross the dataset. To keep things simpler, I'll split the dataset by marker.

```python
AT8_df = MTG_quant[[c for c in MTG_quant.columns if 'AT8' in c]]
pTDP43_df = MTG_quant[[c for c in MTG_quant.columns if 'pTDP43' in c]]
Iba1_df = MTG_quant[[c for c in MTG_quant.columns if 'Iba1' in c]]
AB_df = MTG_quant[[c for c in MTG_quant.columns if '6e10' in c]]
Hema_df = MTG_quant[[c for c in MTG_quant.columns if 'Hematoxylin' in c]]
aSyn_df = MTG_quant[[c for c in MTG_quant.columns if 'aSyn' in c]]
NeuN_df = MTG_quant[[c for c in MTG_quant.columns if 'NeuN' in c]]
GFAP_df = MTG_quant[[c for c in MTG_quant.columns if 'GFAP' in c]]
```

The purpose of the following analysis is to assess which marker will be most effective at building a predictive model for ApoE genotype.  To this end, I will breifly go through each df to look for evidence of a correltation between marker prevelance and ApoE risk alleles.

```python
list(Hema_df.columns)
```

```python
Hema = pd.concat([Hema_df.iloc[:, 7:13], Hema_df.iloc[:, 19:25]], axis=1)
```

```python
AB = pd.concat([AB_df.iloc[:, 6:12], AB_df.iloc[:, 18:24], AB_df.iloc[:, 30:36]], axis=1)
AB['Donor ID'] = MTG_quant['Donor ID']
AB = pd.merge(meta_short, AB, how='inner', on='Donor ID')
AB.head()
```

```python
numeric_cols = AB.select_dtypes(include='number').columns
AB_apoe = AB.groupby('APOE Genotype')[numeric_cols].mean()
AB_apoe
```

```python
fig, axes = plt.subplots(5, 4, figsize=(15, 15))

# Flatten axes for easy looping
axes = axes.flatten()
vars_to_plot = list(AB_apoe.columns)
for ax, v in zip(axes, vars_to_plot):
    sns.barplot(data=AB_apoe, x='APOE Genotype', y=v, ax=ax)
    ax.set_title(f'{v} vs ApoE Genotype', fontsize=8)
    ax.set_ylabel(v, fontsize=6)
    ax.set_xlabel("ApoE Genotype", fontsize=6)
    ax.tick_params(axis='both', labelsize=6)

# If there are empty cells, hide them
for i in range(len(vars_to_plot), len(axes)):
    axes[i].set_visible(False)

plt.tight_layout()
plt.show()
```

```python
AT8 = pd.concat([AT8_df.iloc[:, 13:19], AT8_df.iloc[:, 37:43], AT8_df.iloc[:, 19:25]], axis=1)
pTDP43 = pd.concat([pTDP43_df.iloc[:, 13:19], pTDP43_df.iloc[:, 31:43]], axis=1)
aSyn = pd.concat([aSyn_df.iloc[:, 7:13], pTDP43_df.iloc[:, 19:31]], axis=1)
NeuN = pd.concat([NeuN_df.iloc[:, 7:13], NeuN_df.iloc[:, 37:67]], axis=1)
Iba1 = pd.concat([Iba1_df.iloc[:, 7:13], Iba1_df.iloc[:, 25:49]], axis=1)
GFAP = pd.concat([GFAP_df.iloc[:, 7:25]], axis=1)
Hema = pd.concat([Hema_df.iloc[:, 7:13], Hema_df.iloc[:, 19:25]], axis=1)

markers = {'AT8':AT8, 'pTDO43':pTDP43, 'aSyn':aSyn, 'NeuN':NeuN, 'Iba1':Iba1, 'GFAP':GFAP, 'Hema':Hema}
apoe_results = {}

for name, df in markers.items():
    df['Donor ID'] = MTG_quant['Donor ID']
    df = pd.merge(meta_short, df, how='inner', on='Donor ID')
    numeric_cols = df.select_dtypes(include='number').columns
    apoe_results[name] = df.groupby('APOE Genotype')[numeric_cols].mean()

```

```python
for name, df in apoe_results.items():

    fig, axes = plt.subplots(5, 4, figsize=(15, 15))

    # Flatten axes for easy looping
    axes = axes.flatten()
    vars_to_plot = list(df.columns)
    for ax, v in zip(axes, vars_to_plot):
        sns.barplot(data=df, x='APOE Genotype', y=v, ax=ax)
        ax.set_title(f'{v} vs ApoE Genotype', fontsize=8)
        ax.set_ylabel(v, fontsize=6)
        ax.set_xlabel("ApoE Genotype", fontsize=6)
        ax.tick_params(axis='both', labelsize=6)

    # If there are empty cells, hide them
    for i in range(len(vars_to_plot), len(axes)):
        axes[i].set_visible(False)

    plt.tight_layout()
    plt.show()
```
